## Descripción del dataset

El archivo scopus.csv contiene artículos científicos exportados desde Scopus. Incluye metadatos como autores, título, año, fuente, citas, DOI, resumen y tipo de documento.


```python
import pandas as pd
import textwrap
import matplotlib.pyplot as plt
import seaborn as sns
# Load the dataset
df = pd.read_csv('scopus_g2_ventas.csv')
```

## Gráfico 1: Publicaciones por año

Este gráfico muestra la cantidad de artículos científicos publicados por año relacionados con la predicción de ventas mediante inteligencia artificial, inteligencia de negocios y minería de datos. Permite identificar los años con mayor producción científica sobre el tema.
En este caso se observa la predominancia del año 2025 y la cantidad de publicaciones que se hicieron


```python
publicaciones_por_anio = df["Year"].value_counts().sort_index()

plt.figure(figsize=(10, 5))
bars = plt.bar(publicaciones_por_anio.index.astype(str), publicaciones_por_anio.values)

for i, bar in enumerate(bars):
    year = int(publicaciones_por_anio.index[i])
    if year > 2019:
        bar.set_color('steelblue')
    else:
        bar.set_color('skyblue')

plt.title("Distribución de publicaciones científicas por año", fontsize=15, fontweight="bold")
plt.xlabel("Año de publicación")
plt.ylabel("Cantidad de publicaciones")

for bar in bars:
    valor = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        valor,
        str(int(valor)),
        ha="center",
        va="bottom",
        fontweight="bold"
    )

plt.xticks(rotation=45)
plt.grid(axis="y", linestyle="--", alpha=0.3)
plt.tight_layout()
plt.show()
```

## Gráfico 2: Artículos más citados

Este gráfico muestra los 10 artículos con mayor cantidad de citas dentro del dataset de Scopus. Su finalidad es identificar las investigaciones con mayor impacto académico relacionadas con la predicción de ventas mediante inteligencia artificial, inteligencia de negocios y minería de datos.

```python

df["Cited by"] = pd.to_numeric(df["Cited by"], errors="coerce").fillna(0)

top_citados = df.nlargest(10, "Cited by").copy()

top_citados["Titulo"] = top_citados["Title"].apply(
    lambda texto: "\n".join(textwrap.wrap(str(texto), width=60))
)

plt.figure(figsize=(14, 10))

bars = plt.barh(top_citados["Titulo"], top_citados["Cited by"])

plt.title("Top 10 artículos más citados", fontsize=16, fontweight="bold")
plt.xlabel("Cantidad de citas")
plt.ylabel("Artículo científico")

for bar in bars:
    valor = bar.get_width()
    plt.text(
        valor - 0.5,
        bar.get_y() + bar.get_height() / 2,
        str(int(valor)),
        va="center",
        ha="right",
        color="white",
        fontweight="bold"
    )

plt.gca().invert_yaxis()
plt.grid(axis="x", linestyle="--", alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
top_fuentes = df["Source title"].value_counts().head(10).reset_index()
top_fuentes.columns = ["Fuente", "Cantidad"]

top_fuentes["Fuente"] = top_fuentes["Fuente"].apply(
    lambda texto: "\n".join(textwrap.wrap(str(texto), width=55))
)

plt.figure(figsize=(14, 9))

bars = plt.barh(top_fuentes["Fuente"], top_fuentes["Cantidad"])

plt.title("Top 10 fuentes científicas más frecuentes", fontsize=16, fontweight="bold")
plt.xlabel("Cantidad de publicaciones")
plt.ylabel("Fuente científica")

for bar in bars:
    valor = bar.get_width()
    plt.text(
        valor - 0.05,
        bar.get_y() + bar.get_height() / 2,
        str(int(valor)),
        va="center",
        ha="right",
        color="white",
        fontweight="bold"
    )

plt.gca().invert_yaxis()
plt.grid(axis="x", linestyle="--", alpha=0.3)
plt.tight_layout()
plt.show()
```

## Gráfico 4: Distribución por tipo de documento

Este gráfico de torta muestra la distribución de las publicaciones científicas por tipo de documento. Permite visualizar qué tipos de documentos (por ejemplo, artículos, revisiones, ponencias de congresos) son más frecuentes en el dataset.

```python
import numpy as np
publicaciones_por_tipo = df['Document Type'].value_counts()
colors = plt.cm.Blues(np.linspace(0.3, 1, len(publicaciones_por_tipo)))

plt.figure(figsize=(10, 8))
plt.pie(publicaciones_por_tipo, labels=publicaciones_por_tipo.index, autopct='%1.1f%%', startangle=90, colors=colors)
plt.title('Distribución por tipo de documento', fontsize=16, fontweight='bold')
plt.axis('equal')
plt.tight_layout()
plt.show()
```
